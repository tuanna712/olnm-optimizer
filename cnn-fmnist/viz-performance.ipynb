```python
import json
with open('results/opt_fashion_results_sgd.json') as f:
    results_sgd = json.load(f)
with open('results/opt_fashion_results_adam.json') as f:
    results_adam = json.load(f)
with open('results/opt_fashion_results_olnm.json') as f:
    results_olnm = json.load(f)
```

```python
all_results = {**results_olnm, **results_adam, **results_sgd}
all_results.keys()
```

```python
# --- Comparison Plot ---
import numpy as np
def smooth_history(history, window_size=2):
      history = np.array(history)
      kernel = np.ones(window_size) / window_size
      smoothed = np.convolve(history, kernel, mode='valid')
      pad = (window_size - 1) // 2
      return np.pad(smoothed, (pad, window_size - 1 - pad), mode='edge')

criteria = {0:'train_losses', 1:'test_losses', 2:'train_errs', 3:'test_errs', 4:'train_accs', 5:'test_accs', 6:'run_times'}
methods = {0: 'SGD', 1:'Adam', 3:'Adaptive_OLNM', 2:'OLNM'}

import matplotlib.pyplot as plt
plt.figure(figsize=(8, 6))
plt.rcParams['font.family'] = 'Helvetica'

plt.plot(smooth_history(all_results['SGD'][0]), 
        alpha=0.8, color='blue',
        linewidth=2.5, 
        linestyle='dotted',
        label='SGD (LR=0.2)', 
        markevery=10, marker='*')
plt.plot(smooth_history(all_results['Adam'][0]), 
        alpha=0.8, color='red',
        linewidth=2.5, 
        linestyle='dotted',
        label='Adam (LR=0.005)', 
        markevery=10, marker='*')
plt.plot(smooth_history(all_results['OLNM'][0]), 
        alpha=0.8, color='orange',
        linewidth=2.5, 
        linestyle='dotted',
        label='OLNM (LR=0.04)', 
        markevery=10, marker='8')

plt.xlabel('Epochs', fontsize=16)
plt.xticks(fontsize=16)

plt.ylabel('LCE Training Loss', fontsize=16)
plt.yticks(fontsize=16)
# plt.ylim(0.0001, 0.0015)

plt.legend(loc='upper right', fontsize=16)
plt.grid(True, alpha=0.5)
plt.savefig("imgs/png/figure-5a-fmnist.png")
plt.savefig("imgs/pdf/figure-5a-fmnist.pdf", bbox_inches='tight')
plt.savefig("imgs/svg/figure-5a-fmnist.svg", bbox_inches='tight')
plt.show()
```
